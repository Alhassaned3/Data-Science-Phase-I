 rule 1. Primary Measurement Values Must Be Positive
     verifies that the four primary measurement columns contain only positive values;

```python
primary_measurements = [
    "culmen_length_mm",
    "culmen_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

try:
    assert (clean_df[primary_measurements] > 0).all().all(), "Measurements are not all positive."
    print("Assertion passed: All primary measurements are positive.")
except AssertionError as e:
    print(f"Assertion failed: {e}")
```

Rule 2. Allowed Species Categories
    verifies that species column contains only Adelie, Chinstrap, and Gentoo;

```python
allowed_species = ["Adelie", "Chinstrap", "Gentoo"]

try: 
    assert clean_df["species"].unique().isin(allowed_species).all(), "Species recorded is not allowed"
    print ("Assertion passed: All Recorded Species are allowed")
except AssertionError as e: 
    print(f"Assertion failed: {e}")    
```

Rule 3: Allowed Island Categories
    verifies that island contains only the categories observed in the original dataset;

```python
allowed_island= ['Torgersen', 'Biscoe', 'Dream']

try:
    assert clean_df ["island"].unique().isin(allowed_island).all(), "Contains island that doesnt belong in data set"
    print ("Island Column contains only the categories observed in the original dataset")
except AssertionError as e:
    print(f"Assertion failed: {e}")
```

Rule 4: No Missing Values in Required Fields
    verifies that the required fields from Task 3 contain no missing values 


```python
## check that required fields have no missing value
try: 
    assert clean_df[required_columns].notna().all().all(),"Dataset contains missing values"
    print("Data set contains no missing values")
except AssertionError as e: 
    print(f"Assertion failed:{e}")
```

Rule 5: No Exact Duplicate Rows
    verifies that the cleaned dataset contains no exact duplicate rows.

```python
## verify no duplicates
try:
    assert clean_df.duplicated().sum() == 0,"Duplicate rows detected."
    print("Dataset contains no duplicated entries")
except AssertionError as e:
    print(f"Assertion failed:{e}")
```

### B. Group summaries
    Calculate and display:
    Use groupby() and use agg() for the summaries requiring multiple statistics or columns. Round
    displayed means to two decimal places.

1. the number of penguins in each species;

```python
clean_df.groupby("species").agg(
    penguin_size = ("species", "size")
)
```

Used groupby() to group penguins by species name, then used agg() with size to find the total in each group.  
The data shows Adelie is the most frequent species in this dataset, making up almost half of the total population sampled.

2. the mean body mass for each species;

```python
clean_df.groupby("species").agg(
    penguin_mass = ("body_mass_g", "mean")
).round(2)
```

Used groupby() to group penguins by species name, then agg() with mean to find the mean of each group and added round() to round to the second decimal point.  
The data shows Gentoo penguins are significantly heavier than the other two species

3. the mean culmen length, mean flipper length, and mean body mass for each species; 

```python
clean_df.groupby("species").agg(
mean_culmen_length=("culmen_length_mm", "mean"),
mean_flipper_length=("flipper_length_mm", "mean"),
mean_body_mass_g=("body_mass_g", "mean"),
).round(2)
```

Used groupby() to group penguins by species name, then agg() with mean to calculate the mean culmen length, mean flipper length, and mean body mass for each species.  
The data shows that flipper length increases as mass increases.

4. the number and mean body mass of penguins grouped by both species and sex.


```python
clean_df.groupby(["species","sex"]).agg(
    penguin_count = ("species","size"),
    mean_body_mass_g = ("body_mass_g", "mean"),
).round(2)
```

Used groupby(["species", "sex"]) to split the dataset by both species and sex, then used .agg with size and species to find to total count of penguins that fall under each sex for each of the different species. I then did the same but used mean to caluculate the mean body mass, ending with round() to round each mean to the 2 decimal point. \
The data shows to be well balanced between mall and female test, male seem to consistenly be heavier than there femal counterparts.
