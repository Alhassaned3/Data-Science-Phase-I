```python
## create copy of data file 
raw_df = df.copy()
clean_df = df.copy()
```

```python
## rename files 
clean_df = clean_df.rename(columns = {
    "studyName": "study_name",
    "Sample Number": "sample_number",
    "Species" : "species",
    "Island" : "island",
    "Clucth Completion" : "clucth_completion",
    "Date Egg": "date_egg",
    "Individual ID" : "individual_id",
    "Culmen Length (mm)" : "culmen_length_mm",
    "Culmen Depth (mm)" : "culmen_depth_mm",
    "Flipper Length (mm)" : "flipper_length_mm",
    "Body Mass (g)" : "body_mass_g",
    "Sex" : "sex",
    "Region": "region",
    "Stage": "stage",
    "Delta 15 N (o/oo)": "delta_15_n",
    "Delta 13 C (o/oo)": "delta_13_c",
    "Comments": "comments"
})
```

```python
## convert date to pandas data type
clean_df ["date_egg"] = pd.to_datetime(
    clean_df["date_egg"],
    errors = "coerce" ## if can't convert, don't raise error
                                    )
```

```python
## convert length to numeric data type 
clean_df ["culmen_length_mm"] = pd.to_numeric( 
    clean_df["culmen_length_mm"],
        errors = "coerce"
    )
```

```python
## convert depth to numeric data type 
clean_df ["culmen_depth_mm"] = pd.to_numeric( 
    clean_df["culmen_depth_mm"],
        errors = "coerce" )
```

```python
## convert flipper to numeric data type 
clean_df ["flipper_length_mm"] = pd.to_numeric( 
    clean_df["flipper_length_mm"],
        errors = "coerce" )
```

```python
## convert mass to numeric data type 
clean_df ["body_mass_g"] = pd.to_numeric( 
    clean_df["body_mass_g"],
        errors = "coerce" )
```
